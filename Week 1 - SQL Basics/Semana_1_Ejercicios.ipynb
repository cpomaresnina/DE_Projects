```sql
-- Cuantos viajes hay en total en el datase de NYC Taxi trips
SELECT count(*) from samples.nyctaxi.trips;
```

```sql
-- Que columnas tiene la tabla nyctaxi y q tipo de datos tiene?
DESC TABLE samples.nyctaxi.trips;
```

```sql
-- Cual es el viajes mas largo y el mas corto, exlcluyendo ceros
SELECT MAX(trip_distance), MIN(trip_distance) FROM samples.nyctaxi.trips WHERE trip_distance > 0
;
```

```sql
-- Cual es el promedio de tarifa y distancia de los viajes
SELECT round(avg(fare_amount),2), round(avg(trip_distance),2) FROM samples.nyctaxi.trips;
```

```sql
-- Muestra los 20 viajes mas recientes con su fecha distancia y tarifa
SELECT * FROM samples.nyctaxi.trips ORDER BY tpep_pickup_datetime DESC LIMIT 20;
```

```sql
-- Muestra todos lo viajes q tienen una distancia mayor a 10 millas
SELECT * from samples.nyctaxi.trips WHERE trip_distance > 10;
```

```sql
-- VIAJES CON DISTANCIA MAYOS A 5 Y TARIFA MAYOR A 10
SELECT * FROM samples.nyctaxi.trips
WHERE trip_distance>5 AND fare_amount>10;
```

```sql
-- Viajes con tyarifa entrte 10 y 20
SELECT * FROM samples.nyctaxi.trips WHERE fare_amount BETWEEN 10 AND 20
```

```sql
-- Viajes q comenzaron en zip 10009, 10013, 10014
SELECT * FROM samples.nyctaxi.trips WHERE pickup_zip IN (10009,10013,10014)
```

```sql
-- total de tarifas y distancias de viajes
SELECT SUM(fare_amount), sum(trip_distance) from samples.nyctaxi.trips;
```

```sql
-- distancia y tarifa proimedio
SELECT AVG(fare_amount) as tarifa_promedio, avg(trip_distance) as distrancia_promedio from samples.nyctaxi.trips WHERE trip_distance > 0;
```

```sql
SELECT MIN(fare_amount), MIN(trip_distance), MAX(fare_amount), MAX(trip_distance) from samples.nyctaxi.trips
where trip_distance>0 and fare_amount >0
```

```sql
-- 5.1
SELECT pickup_zip, COUNT(*) AS total_viajes FROM samples.nyctaxi.trips
GROUP BY pickup_zip
having total_viajes >1000
ORDER BY total_viajes DESC
;

```

```sql
-- 5.2
SELECT date(tpep_pickup_datetime), count(*) as total_viajes, avg(fare_amount)
from samples.nyctaxi.trips
group by date(tpep_pickup_datetime)
having total_viajes > 200
order by total_viajes DESC
```

```sql
--5.3
SELECT HOUR(tpep_pickup_datetime), count(*) as total_viajes
FROM samples.nyctaxi.trips
GROUP BY HOUR(tpep_pickup_datetime)
HAVING total_viajes > 1000
ORDER BY total_viajes DESC
```

```sql
-- 6.1
SELECT pickup_zip, count(*) as total_viajes, max(fare_amount), avg(fare_amount), min(fare_amount)
FROM samples.nyctaxi.trips
WHERE trip_distance > 0 AND fare_amount>5
group by pickup_zip
HAVING total_viajes > 1000
ORDER BY total_viajes DESC

```

```sql
--6.2
SELECT 
    CASE
        WHEN trip_distance<2 THEN 'Corto'
        WHEN trip_distance between 2 and 5 THEN 'Medio' 
        WHEN trip_distance between 5 and 10 then 'Largo'
        ELSE "Muy Largo" 
    END as Categoria, 
    count(*) as total_viajes,
    avg(fare_amount) as promedio_tarifa,
    avg(trip_distance) as distancia_promedio
FROM samples.nyctaxi.trips
GROUP BY Categoria
ORDER BY total_viajes DESC
```
