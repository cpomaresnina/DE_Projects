```sql
--2.1 
select c.cliente_id, c.nombre, c.ciudad, o.orden_id, o.producto_id, o.cantidad, o.fecha
from bootcamp.practice.clientes c
JOIN bootcamp.practice.ordenes o
ON c.cliente_id = o.cliente_id
order by c.cliente_id
```

```sql
-- 2.2
select o.orden_id, o.fecha, p.nombre, p.categoria, p.precio, o.cantidad * p.precio as recaudacion
from bootcamp.practice.ordenes o
join bootcamp.practice.productos p using(producto_id)
order by orden_id
```

```sql
--2.3
SELECT c.nombre, c.ciudad, o.orden_id, o.fecha, p.nombre, p.categoria, p.precio, o.cantidad * p.precio as total
FROM bootcamp.practice.clientes c
join bootcamp.practice.ordenes o using (cliente_id)
join bootcamp.practice.productos p using (producto_id)
```

```sql
-- 6.1
select e.empleado_id, e.nombre, e.cargo, e.jefe_id, e2.nombre as jefe
from bootcamp.practice.empleados e
join bootcamp.practice.empleados e2 on e.jefe_id=e2.empleado_id

```

```sql
-- 6.2
select e.empleado_id, e.nombre, e.cargo, e.jefe_id, e2.nombre as jefe
from bootcamp.practice.empleados e
left join bootcamp.practice.empleados e2 on e.jefe_id=e2.empleado_id
where e.jefe_id is null

```

```sql
--6.3
SELECT 
    j.nombre as jefe,
    j.cargo as cargo_jefe,
    e.nombre as subordinado,
    e.cargo as cargo_subordinado
FROM bootcamp.practice.empleados e
JOIN bootcamp.practice.empleados j ON e.jefe_id = j.empleado_id
ORDER BY j.nombre, e.nombre;
```

```sql
-- 7.1
select c.nombre, c.ciudad, o.orden_id, o.fecha, p.nombre, p.categoria, o.cantidad, p.precio, o.cantidad*p.precio as total 
from bootcamp.practice.ordenes o
inner join bootcamp.practice.clientes c using(cliente_id)
inner join bootcamp.practice.productos p using(producto_id)
order by o.fecha desc, c.nombre asc
```

```sql
-- 7.1
select c.nombre, c.ciudad, count(o.orden_id), sum(p.precio*o.cantidad) as total
from bootcamp.practice.ordenes o
inner join bootcamp.practice.clientes c using(cliente_id)
inner join bootcamp.practice.productos p using(producto_id)
group by c.nombre, c.ciudad
```

```sql
--7.3
SELECT 
    p.nombre as producto,
    p.categoria,
    COALESCE(SUM(o.cantidad), 0) as cantidad_total_vendida,
    COALESCE(SUM(o.cantidad * p.precio), 0) as ingresos_totales
FROM bootcamp.practice.productos p
LEFT JOIN bootcamp.practice.ordenes o ON p.producto_id = o.producto_id
GROUP BY p.producto_id, p.nombre, p.categoria
ORDER BY ingresos_totales DESC;
```
